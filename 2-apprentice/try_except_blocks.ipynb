# Try and Except Blocks

Run the cells from top to bottom. Outputs are shown directly below the examples that produce them.


```python
try:     #place code which you expect can throw an exception in this block
    print(variable)     #if no exception is triggered by any of the code, the code execution proceeds past the except block

except:     #if an exception is raised, then the exception output is not printed out to the console. Instead the code within the except block is executed
    print("Exception occurred(Name Error), because 'variable' was not defined ")
```

```python
try:
    print(variable)

except NameError:
    print("Variable is not defined")

except:    #for any other exception, this general purpose except block is used
    print("an unknown error has occurred")
```

```python
try:
    f = open("nonexistent_file")

except FileNotFoundError:
    print('No such file')

except:
    print('An unknown error has occurred')
```

```python
try:
    f = open("nonexistent_file")
except OSError:
    print("OS Error")   #This except block is ran
except FileNotFoundError:
    print('File not found Error')
```

```python
try:
    f = open("nonexistent_file")
except FileNotFoundError:
    print('File not found Error')  #this except block is ran
except OSError:
    print("OS Error")
```

```python
try:
    print(var)
except NameError:
    print("The var has not been defined")
finally: #finally block is always executed
    print("The 'try except block' ended")
```

```python
try:
    file = open('testfile.txt', 'w')

finally:
    print ("Done with processing")
```

```python
print(file.closed) #False
```

```python
try:
    file = open("testfile.txt")

finally:
    file.close()
    print("Done with processing the file.")
```

```python
try:
    file = open('testfile.txt')
    file.write("Writing to the open file")

except:
    print("Ooops! Something went wrong")

else:
    print("Nothing went wrong")
```

```python
try:
    file = open('testfile.txt', 'w')
    file.write("Writing to the open file")
    print("One line written to the file")

except:
    print("Ooops! Something went wrong")

else:
    print("Nothing went wrong")
```
