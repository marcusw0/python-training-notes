# Stack

A stack is a **last in, first out (LIFO)** data structure. The newest item added is the first item removed, like a browser back button, undo history, or a call stack.


```python
stack = []
```

## Using a Python list as a stack

Python lists already support the two core stack operations: `append()` pushes onto the top, and `pop()` removes from the top. Both operations are efficient at the end of the list.


```python
stack.append('USA')
stack.append('GBR')
stack.append('CHN')
```

## Custom stack class

This wrapper gives the stack a clear API: `push`, `pop`, `peek`, `is_empty`, and `size`. The list is still doing the storage work, but the class makes the data structure's rules explicit.


```python
class stack:

    def __init__(self):
        """Create new stack"""
        self.stack = []

    def push(self, data):
        """Add an element to the stack"""
        self.stack.append(data)

    def pop(self):
        """Remove the top element of the stack"""
        if self.is_empty():
            raise Exception("nothing to pop")
        
        return self.stack.pop(len(self.stack) - 1)
    
    def peek(self):
        """Have a look at the top element of the stack"""
        if self.is_empty():
            raise Exception("Nothing to peek")
        
        return self.stack[len(self.stack) - 1]
    
    def is_empty(self):
        """Returns true if stack is empty"""
        return len(self.stack) == 0
    
    def size(self):
        """Returns size of stack"""
        return len(self.stack)
```

```python
olympics = stack()
olympics
```

```python
olympics.push('USA')
olympics.push('GBR')
olympics.push('CHN')
olympics.push('RUS')
```

```python
olympics.size()
```

```python
olympics.pop()
```

```python
olympics.is_empty()
```

```python
olympics.peek()
```

```python
olympics.pop()
```

## Empty-stack behavior

The final calls remove every value. The next `pop()` is expected to raise an exception, which demonstrates **stack underflow**: trying to remove data when no data exists.


```python
olympics.pop()
olympics.pop()
```
