# Graph Traversal

Traversal is the process of visiting reachable vertices from a starting point. BFS and DFS visit the same graph with different ordering strategies.


```python
from graph import *

from queue import Queue
import numpy as np
```

## Breadth-first search

BFS uses a queue. It visits all immediate neighbors before moving farther away, making it useful for shortest paths in unweighted graphs and level-order exploration.


```python
def breadth_first(graph, start=0):

    queue = Queue()
    queue.put(start)

    visited = np.full((graph.num_vertices, ), False, dtype=bool)

    while not queue.empty():
        vertex = queue.get()

        if visited[vertex]:
            continue

        print("Visited: ", vertex)
        visited[vertex] = True

        for v in graph.get_adjacent_vertices(vertex):
            if not visited[v]:
                queue.put(v)
```

## Example graph

The graph below has nine vertices and undirected edges. Starting BFS at vertex `2` shows which vertices are discovered from that point.


```python
g = AdjacencyMatrixGraph(9)

g.add_edge(0, 1)
g.add_edge(0, 2)
g.add_edge(2, 5)
g.add_edge(2, 4)
g.add_edge(2, 3)
g.add_edge(1, 5)
g.add_edge(5, 6)
g.add_edge(7, 3)
g.add_edge(3, 4)
g.add_edge(2, 5)

g.show()
```

```python
breadth_first(g, 2)
```

## Depth-first search

DFS uses recursion here. Each call visits one vertex, then recursively explores each unvisited neighbor before backtracking.


```python
def depth_first(graph, visited, current=0):
    if visited[current]:
        return
    
    visited[current] = True

    print("Visited: ", current)

    for vertex in graph.get_adjacent_vertices(current):
        depth_first(graph, visited, vertex)
```

```python
visited = np.full((g.num_vertices, ), False, dtype=bool)

depth_first(g, visited)
```
