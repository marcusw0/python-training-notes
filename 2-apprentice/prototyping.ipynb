# Prototyping

Run the cells from top to bottom. Outputs are shown directly below the examples that produce them.


```python
num_list = [10,20,30,40,50]
```

```python
def main():

    count = 0
    total = 0

    for i in num_list:
        count = count + 1
        total = total + i
##below is for debugging
    print('total sum', total)
```

```python
if __name__ =="__main__":
    main()     #will be invoked only if the current source is being invoked directly rather than being imported into another source, and then being invoked from there
```
