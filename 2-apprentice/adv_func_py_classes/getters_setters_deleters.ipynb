```python
class Wrestler:

    def __init__(self):
        self.__name = ''

    def set_name(self, name):
        print('setter method called')
        self.__name = name

    def get_name(self):
        print('getter method called')
        return self.__name
    
    name = property(get_name, set_name)
    #property(fget, fset, fdel, doc)
    ##fget - getter function, fset - setter function, fdel - deleter function, doc - documentation
```

```python
w = Wrestler()
w.name = 'Kart'
```

```python
w.name
```

```python
class Wrestler:

    def __init__(self):
        self.__name = ''

    def set_name(self, name):
        print('setter method called')
        self.__name = name

    def get_name(self):
        print('getter method called')
        return self.__name
    
    def del_name(self):
        print('deleter method called')
        del self.__name
    
    name = property(get_name, set_name, del_name)
```

```python
w = Wrestler()
w.name = 'Kane'
```

```python
w.name
```

```python
del w.name
```

```python
w.name
```
