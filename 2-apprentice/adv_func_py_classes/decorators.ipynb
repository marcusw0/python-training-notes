```python
class Wrestler:

    def __init__(self, name):
        self.__name = name

    #decorator
    @property #tells python this is the method you want to invoke when the name property is accessed
    def name(self):
        print('getter method called')
        return self.__name
    
    @name.setter #setter func to be invoked on name property
    def name(self, value):
        print('setter method called')
        self.__name = value

    @name.deleter
    def name(self):
        print('deleter method called')
        del self.__name
```

```python
w = Wrestler('Adam')
w.name
```

```python
w.name = 'John'
```

```python
del w.name
```

```python
w.name
```

```python
class Wrestler:

    def __init__(self, name, age):
        self.__name = name
        self.__age = age
    
    @property
    def name(self):
        print('name getter method called')
        return self.__name
    
    @name.setter #setter func to be invoked on name property
    def name(self, value):
        print('name setter method called')
        self.__name = value

    @name.deleter
    def name(self):
        print('name deleter method called')
        del self.__name

    @property
    def age(self):
        print('age getter method called')
        return self.__age
    
    @age.setter
    def age(self, value):
        print('age setter method called')
        self.__age = value

    @age.deleter
    def age(self):
        print('age deleter method called')
        del self.__age
```

```python
w = Wrestler('Mark', 25)
```

```python
w.name
```

```python
w.name = 'Joe'
```

```python
del w.name
```

```python
w.age
```

```python
w.age = 26
```

```python
del w.age
```
