```python
from abc import ABC, abstractmethod
```

```python
class Hominidae():

    def diet(self):
        pass

    def walk(self):
        pass

    def behavior(self):
        print("They show complex facial expression and social behaviour.")
```

```python
chimpanzee = Hominidae()
chimpanzee.behavior()
```

```python
chimpanzee.diet()
chimpanzee.walk()
```

```python
class Human(Hominidae):

    def diet(self):
        print("Humans are omnivorous.")

    def walk(self):
        print("They are bipeds.")
```

```python
paul = Human()
paul.diet()
paul.walk()
```

```python
class Hominidae(ABC):

    @abstractmethod
    def diet(self):
        pass

    def walk(self):
        pass

    def behavior(self):
        print("They show complex facial expression and social behaviour.")
```

```python
great_apes = Hominidae()
```
