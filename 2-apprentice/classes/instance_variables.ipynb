```python
class Competition:
    def __init__(self, name, prize):
        self.name = name
        self.prize = prize
        self.participants = []
```

```python
debate = Competition('Debate', 500)
debate.participants.append('Alice')
print(debate.participants)
```

```python
essay = Competition('Essay', 456)
essay.participants.append("Michael")
print(essay.participants)
print(debate.participants)
```
