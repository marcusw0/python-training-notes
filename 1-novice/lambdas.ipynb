# Lambdas

Run the cells from top to bottom. Outputs are shown directly below the examples that produce them.


```python
cube_of = lambda x: x * x * x
```

```python
add = lambda x, y: x + y
```

```python
sub = lambda x, y: x - y
```

```python
mul = lambda x, y: x * y
```

```python
div = lambda x,y: x / y
```

```python
def calculate(x, y, operation="add"):

    if operation == "add":
        return add(x,y)
    
    if operation == 'sub':
        return sub(x,y)
    
    if operation == 'mul':
        return mul(x,y)
    
    if operation == 'div':
        return div(x,y)
```

```python
print(calculate(100, 4, operation="sub"))
```

```python
print(calculate(100, 4))
```

```python
def calculate(*args, operation="add"):

    if operation == "add":
        return add(*args)
    
    if operation == 'sub':
        return sub(*args)
    
    if operation == 'mul':
        return mul(*args)
    
    if operation == 'div':
        return div(*args)
```

```python
print(calculate(100,4, operation="sub"))
```

```python
def calculate(*args, operation=add):
    return operation(*args)
```

```python
print(calculate(100, 4, operation=sub))
```

```python
calculate(10,4, operation=lambda x, y: x**y)
```

```python
print(calculate(10,4, operation=lambda x, y: x**y))
```

## filter()


```python
num_list = [1,5,6,7,11,78,99,34,105,214]
```

```python
filter(lambda x: x > 10, num_list)
```

```python
greater_than_10_list = list(filter(lambda x: x > 10, num_list))
```

```python
print(greater_than_10_list)
```

```python
even_list = list(filter(lambda x: x %2 == 0, num_list))
```

```python
print(even_list)
```

```python
greater_than_10_less_than_100 = list(filter(lambda x: x > 10 and x < 100, num_list))
```

```python
print(greater_than_10_less_than_100)
```
