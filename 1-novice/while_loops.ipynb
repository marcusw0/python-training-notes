```python
c_degrees = range(-20, 45, 5)
```

```python
c_degrees = list(c_degrees)
```

```python
index = 0
```

```python
print('    C    F')
```

```python
while index < len(c_degrees):

    C = c_degrees[index]
    F = (9.0/5)*C + 32

    print('%5d %5.1f' % (C,F))

    index += 1
```

```python
clean_haystack = ['hay','hay','hay','hay',
                  'hay','hay','hay','hay']
```

```python
unclean_haystack = ['hay','hay','hay','hay',
                    'hay','needle','hay','hay']
```

```python
searching_for = 'needle'
```

```python
maximum = len(clean_haystack)
```

```python
i = 0
```

```python
print("searching clean_haystack")
```

```python
while i < maximum:

    if searching_for == clean_haystack[i]:
        print("The %s is at the index %i:" % (searching_for, i))

        del clean_haystack[i]
        break

    i += 1

else:
    print("\nThe needle was not found :(")
```

```python
maximum = len(unclean_haystack)
```

```python
i = 0
```

```python
print("searching unclean_haystack")
```

```python
while i < maximum:

    if searching_for == unclean_haystack[i]:
        print("The %s is at the index %i :)" % (searching_for, i))

        del unclean_haystack[i]
        print("The updated haystack is:", unclean_haystack)
        break

    i += 1

else:
    print("\nThe needle was not found :(")
```
