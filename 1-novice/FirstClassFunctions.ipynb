# First Class Functions

Run the cells from top to bottom. Outputs are shown directly below the examples that produce them.


```python
def add(a,b):
    return a + b
```

```python
def sub(a,b):
    return a - b
```

```python
def mul(a,b):
    return a * b
```

```python
def div(a,b):
    return a / b
```

```python
def get_function(operator='+'):

    if operator == '+':
        return add
    
    if operator == '-':
        return sub
    
    if operator == '*':
        return mul
    
    if operator == '/':
        return div
```

```python
func = get_function()
```

```python
print(func(3,4))
```

```python
func = get_function('*')
```

```python
print(func(3,4))
```

```python
calc_dictionary = {
    '+': add,
    '-': sub,
    '*': mul,
    '/': div
}
```

```python
func = calc_dictionary['/']
```

```python
print(func(12,4))
```

```python
print(calc_dictionary['*'](100, 3))
```
