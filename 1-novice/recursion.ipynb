```python
def decrement(num):

    if num == 0:   #terminating condition
        return
    
    print(num)
    decrement(num - 1)
```

```python
def recursive_sum(num):

    if num == 0:
        return 0
    
    result = num + recursive_sum(num - 1)

    return result
```

```python
def fibonacci(number, fib_series):

    if number < 2:
        return
    
    l = len(fib_series)

    new_number = fib_series[l - 1] + fib_series[l - 2]

    fib_series.append(new_number)

    fibonacci(number - 1, fib_series)
```

```python
series = [0, 1]
```

```python
fibonacci(10, series)
```
