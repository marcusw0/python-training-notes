## copying###


```python
import copy
```

```python
automakers = copy.deepcopy(list_str) #creates a copy of list_str inside the new variable automakers
```

```python
print(automakers) #deepcopy creates a copy of the list and all of its elements - any new changes to list_str doesn't affect automakers
```

```python
del list_str #deletes list_str from memory
```

```python
automakers1 = automakers #shallow copy. Both variables refer to the same list copy. Any changes reflect on both
```

```python
print(automakers1)
```

```python
print(automakers)
```

```python
automakers2 = copy.copy(automakers) #Shallow copy but changes aren't reflected on both
```

```python
print('\n')
```

```python
automakers.remove('Toyota')
```

```python
print(automakers)
```

```python
print(automakers1)
```

```python
print(automakers2)
```

## slicing lists ###


```python
my_cars = ['Tacoma','CRX','WRX','Civic','GMC Truck']
```

```python
print(my_cars[0:3])
```

```python
print(my_cars[:3])  #starts at beginning of the list and goes up to (but not including) index 3
```

```python
print(my_cars[3:]) #Starts with index 3 and goes to the end of the list
```

```python
print(my_cars[::2])
```

```python
print(my_cars[0:4:2])
```

```python
print(my_cars[::3])
```

```python
print(my_cars[::-1]) #simple command to reverse the order
```

```python
print(len(my_cars)) #5 in the list
```

```python
my_cars[5:-1] = ['Ford Focus','Dodge Charger'] #not used much but adds these elements to the end of the list
```

```python
print(my_cars)
```

```python
print('Ford Focus' in my_cars)
```

```python
my_cars.pop(-2) #removes Ford Focus
```

```python
print('Ford Focus' in my_cars) #This is now False
```

## Nested lists (matrix) ###


```python
car_matrix = [['Hennessey Venom GT', 1244],
              ['SSC Ultimate Aero', 1287],
              ['Zenvo ST1', 1100]]
```

```python
print(car_matrix)
```

```python
len(car_matrix) #will return the length of the outer lists
```

```python
len(car_matrix[1]) #index 1 of the outer most list. Will calculate the length of the inner list of index 1
```

```python
car_matrix[1][0] #will return SSC Ultimate Aero
```

```python
student_list_4 = [[1, 'Caleb'],
                  [2, 'Leo'],
                  [3, 'Easton'],
                  [4, ['Jordan', 'Adam']]]
```
