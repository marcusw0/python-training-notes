## Formatting Operations with strings


```python
var1 = "barks"
```

```python
print("A dog " + var1 +"!")
```

```python
print(f'A dog {var1}!') #if you specify your string using the f prefix, python will interpret anything within curly braces inside your string as an expression
```

```python
print("A dog {}!".format(var1))
```

```python
print("Hi \t 'How are you?'")
```

```python
print("Hi\t'How are you?'")
```

```python
print('C:\\Users\\Q') #the first backslash will tell python to interpret the second backslash literally
```

## Comprehensions ###


```python
num = range(10)
```

```python
test = [x * x for x in num]
```

```python
print(test)
```

```python
test = [x * x for x in num if x % 2 != 0]
```

```python
print(test)
```

```python
letters = [letter for letter in "anxiety"]
```

```python
print(letters)
```
