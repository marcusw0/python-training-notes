# Closures

Run the cells from top to bottom. Outputs are shown directly below the examples that produce them.


```python
def nested_hello_fn():

    def hello():
        print("Hello Cathy!")

    hello()
```

```python
nested_hello_fn()
```

```python
def get_hello_fn():

    def hello():
        print("Hello Cathy!")

    return hello
```

```python
hello_fn = get_hello_fn()
```

```python
hello_fn()
```

```python
def greet_hello_by_name(name):

    def hello():
        print("Hello!", name)

    hello()

    return hello
```

```python
greet_hello_fn = greet_hello_by_name("Chris")
```

```python
greet_hello_fn()
```

```python
def greet_by_name(name):

    greeting_msg = "Hi there!"

    def greeting():
        print(greeting_msg, name)

    return greeting
```

```python
greet_fn = greet_by_name("Alex")
```

```python
greet_fn()
```

```python
del greet_by_name
```

```python
greet_fn() #still works and still has access to local variables defined in the outer function even though it doesn't exist in Python memory
```

```python
import random
```

```python
random.seed(0)
```

```python
def greet_with_personal_message(name, message):

    annotations = ['-', '*', '+', ':', '^']
    annotate = random.choice(annotations)

    def greeting():
        print(annotate * 50)
        print(message, name)
        print(annotate * 50)

    return greeting
```

```python
greet_greg_fn = greet_with_personal_message('Greg', 'Hello!')
```

```python
greet_amy_fn = greet_with_personal_message('Amy', 'Good Morning!')
```

```python
greet_john_fn = greet_with_personal_message("John", 'Hey!')
```

```python
greet_greg_fn()
```

```python
greet_amy_fn()
```

```python
greet_john_fn()
```

```python
greet_greg_fn() #will always maintain the same annotation
```

```python
del greet_with_personal_message
```

```python
greet_greg_fn()
```

```python
def enroll_in_college(college_name):

    student_list = []

    def enroll_student(student_name):

        student_list.append(student_name)
        print("Student", student_name, "has been enrolled in", college_name)
        print("Current students", student_list, end="\n\n")

    return enroll_student
```

```python
enroll_in_yale_fn = enroll_in_college('Yale')
```

```python
enroll_in_duke_fn = enroll_in_college('Duke')
```

```python
enroll_in_yale_fn("Sam")
```
