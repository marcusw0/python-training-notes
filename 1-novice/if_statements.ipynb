## If-else Elif ###


```python
if 10 > 20:
    print('10 is greater than 20')
    print("if block activated")
else:
    print("10 is less than 20")
    print("else block activated") #if the if block is false, the else block is executed
```

## Ternary operators(conditional expression)


```python
num = 50
```

```python
print("num before expression: ", num)
```

```python
num = num - 20 if num > 20 else num + 20 #num = 30 if 50 > 20. if it isn't num = 70
```

```python
print("num after expression: ", num)
```

## elif


```python
if 15 > 20:
    print("15 is greater than 20")
    print("if block activated")
elif 15 < 20:
    print("15 is less than 20")
    print("elif block activated")
else:
    print("Both are equal")
    print("else block activated")
```

## nested if-else


```python
x = 25
```

```python
y = 35
```

```python
z = 45
```

```python
if x < y:
    print("The first condition is true")

    if x < z:
        print("Both conditions are true")

    else:
        print("The first condition is true, the second is false")
```
