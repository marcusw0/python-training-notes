# For Loops

Run the cells from top to bottom. Outputs are shown directly below the examples that produce them.


## for loops ###


```python
dog_weight = (("Pug", 20),
              ("Doberman", 80),
              ("Golden Retriever", 55))
```

```python
for i, (dog, weight) in enumerate(dog_weight):
    print("The dog  at index %d is a %s and it weighs %s pounds" % (i, dog, weight))
```

```python
student_scores = {'John' : 80, 'Sam' : 60, 'Jill' : 50, 'Bob' : 96}
```

```python
for key in student_scores:
    print("Key vlue pair -", key, ":", student_scores[key])
```

```python
for key_value in student_scores.items():
    print(key_value)
```

```python
for student, score in student_scores.items():
    print("Student:", student, "\tScore:", score)
```

## Ranges ###


```python
items = []
```

```python
for i in range(100, 301):

    num_str = str(i)

    first_digit = int(num_str[0])
    second_digit = int(num_str[1])
    third_digit = int(num_str[2])

    if (first_digit % 2 == 0) and (second_digit % 2 == 0) and (third_digit % 2 == 0):
        items.append(num_str)
```

```python
print(", ".join(items))
```

```python
for num1 in range(1, 10):
    print("num =", num1, ":", end = " ")

    for num_2 in range(1, 10):
        print("{:2d}".format(num1 * num_2), end = " ")

    print()
```

## continue ###


```python
for i in range(1, 10):

    if i % 2 == 0:
        print('Is %s an even number? ' % i)

    else:
        continue

    print('Yes')
```

```python
for num in range(1, 10):

    if num % 2 == 0:
        print("Found an even number:", num)

        continue
    print(num)
```

```python
num_days = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
```

```python
new_dict = dict()
```

```python
for i, j in zip(months, num_days):
    new_dict.update({i:j})

    if new_dict[i] < 29:
        continue

    else:
        print(i, "month has", new_dict[i], "days")
```
