# Generator

Run the cells from top to bottom. Outputs are shown directly below the examples that produce them.


```python
def generator():
    print('one')
    yield 1

    print('two')
    yield 2

    print('three')
    yield 3

    print('four')
    yield 4
```

```python
g = generator()
```

```python
g  #even though the function is called here it is not executed like a normal function would
```

```python
next(g)  #executes all of the code until the first yield
```

```python
next(g) #picks up the control flow of the execution starting after the first yield
```

```python
next(g) #starts after the second yield
```

```python
next(g) #after third yield
```

```python
g = generator()
```

```python
next(g)
```

```python
def generate_even_numbers(limit):

    for i in range(0, limit, 2):
        yield i
```

```python
g = generate_even_numbers(7)
```

```python
print(next(g))
```

```python
print(next(g))
```

```python
print(next(g))
```

```python
print(next(g))
```

```python
g = generate_even_numbers(7)
```

```python
l = list(g)  #list knows it is a generator and invokes next until stop iteration is encountered
```

```python
print(l)
```

## for loop generator function##


```python
def generate_powers_of_two():

    num = 0
    while True:
        num = num +1

        yield 2 ** num
```

```python
gen = generate_powers_of_two()
```

```python
count = 0
```

```python
for p in gen:
    print(p)

    count = count + 1
    if count > 10:
        break
```
