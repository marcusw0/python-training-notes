## args ###


```python
def print_fn(*args):

    args_type = type(args)
    print(args_type)

    print(args)
```

```python
print_fn("bob")
```

```python
print_fn("Greg", "Bob", "Alice")
```

```python
some_list = ["Greg", "Bob", "Alice"]
```

```python
print_fn(some_list)
```

```python
print_fn(*some_list) #unpacks the list
```

```python
def student_details(**kwargs):
    
    print(type(kwargs))
    print(kwargs)
```

```python
student_details()
```

```python
student_details(name="John")
```
