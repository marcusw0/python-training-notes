## zip###


```python
tuple_a = 1,2,3,4,5
```

```python
tuple_b = 'a','b','c','d','e'
```

```python
zipped = zip(tuple_a, tuple_b) # makes pairs of every element in tuple_a with the corresponding element in tuple_b
```

```python
print(zipped) #<zip object at 0x7eff6a6a1300> zips are a collection of tuples. This zip contains (1,'a')(2,'b')(3,'c')(4,'d')(5,'e')
```

```python
result = tuple(zipped) #zips can be converted to any complex data type (tuple,list,dict,set, or any complex data type that python suports)
```

```python
print(result)
```

```python
tuple_x, tuple_y = zip(*result) #will be the original tuples zipped up (a, b) but now stored in x, y
```
