## Invoking functions on Dicts ###


```python
print(bike_details)
```

```python
len(bike_details)
```

```python
len(bike_details.keys())
```

```python
len(bike_details.values())
```

```python
sorted(bike_details)
```

```python
sorted(bike_details, reverse=True) #sorts in the reverse lexicorgraphical order
```

```python
bike_details.items() #will return a list of key value pairs in the form of tuples
```

```python
copy_bike_details = bike_details.copy()
```

```python
print(copy_bike_details)
```

```python
copy_bike_details.pop('num_cylinders') #dicts support pop
```

```python
print(copy_bike_details)
```

```python
print(bike_details) #using pop on the copy did not effect the original
```

```python
copy_bike_details.popitem() # .popitem removes any key value pair from a dict
```

```python
print(copy_bike_details)
```

```python
dict_age = {'Ethan': 54, 'Sofia': 50}
```

```python
print(dict_age)
```

```python
new_dict_age = {'Ethan': 55, 'Harper': 58}
```

```python
print(new_dict_age)
```

```python
dict_age.update(new_dict_age) #updates the dict dict_age with the new value for Ethan (55) and the new key value pair Harper: 58
```

```python
print(dict_age)
```

```python
print(copy_bike_details)
```

```python
print(bike_details)
```

```python
bike_details.clear() #clears all key value pairs and leaves an empty dict
```

```python
print(bike_details)
```

```python
del copy_bike_details #if no longer using, free up all the memory associated with a dict by using del
```
